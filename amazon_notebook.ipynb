I divide one cell by "N" cells by using this rule :
One comment is seperated with another with a if : "comma" and "no-space"
Of course, the last comment won't have a "comma" at the end, so we add it.

```python
def subchain_the_data(str):
    sub_str_list=[]
    index_beginning=0
    for i in range(len(str)-2) :
        if str[i]=="," and str[i+1] !=" ":
            sub_str_list.append(str[index_beginning:i])
            index_beginning=i+1
    sub_str_list.append(str[index_beginning:len(str)])
    return sub_str_list
```

```python
data=pd.read_csv('amazon.csv')
data.head(1)
```

```python
user_id=data["user_id"]
user_name=data["user_name"]
review_id=data["review_id"]
review_title=data["review_title"]
review_content=data["review_content"]
```

```python
#From now on, we divide cells of data, into flattened_array=subchain_the_data(user_id,user_name,review_id, review_title,review_content).
#One can show that a perfect data means : for each row, if len(divided_cells[i]) alls points "i" equal.
#So we try this.
```

```python
length=[]
for i in range(len(data)):
    length.append([len(subchain_the_data(user_id[i])),len(subchain_the_data(user_name[i])),len(subchain_the_data(review_id[i])),len(subchain_the_data(review_title[i])),len(subchain_the_data(review_content[i]))])

correct_rows=[]
incorrect_rows=[]

for i in range(len(data)):
    if length[i][0]==length[i][1]==length[i][2]==length[i][3]==length[i][4] :
        correct_rows.append(i)
    else :
        incorrect_rows.append(i)  
                
correct_data = data.iloc[correct_rows].copy()
```

```python
#We have the data without any problems
#Now we just want to flattend the agregated data.
import re

# Colonnes contenant les données agrégées
colonnes_avis = [
    "user_id",
    "user_name",
    "review_id",
    "review_title",
    "review_content"
]

# Toutes les autres colonnes sont liées au produit
colonnes_produit = [
    col for col in data.columns
    if col not in colonnes_avis
]

# Séparation des valeurs selon ta règle
def separer_donnees(chaine):
    if isinstance(chaine, str):
        return re.split(r",(?=\S)", chaine)
    return [chaine]

# Construction du tableau aplati
lignes = []

for indice, row in data.iterrows():

    # Découpage des cinq colonnes d'avis
    avis = {
        col: separer_donnees(row[col])
        for col in colonnes_avis
    }

    # Vérification du nombre de valeurs
    longueurs = {col: len(avis[col]) for col in colonnes_avis}

    if len(set(longueurs.values())) != 1:
        raise ValueError(
            f"Erreur à la ligne {indice} : "
            f"nombres de valeurs incompatibles : {longueurs}"
        )

    # Une nouvelle ligne par avis
    for i in range(len(avis["user_id"])):

        nouvelle_ligne = {
            col: row[col]
            for col in colonnes_produit
        }

        for col in colonnes_avis:
            nouvelle_ligne[col] = avis[col][i]

        lignes.append(nouvelle_ligne)

# Création du DataFrame final
data_to_csv = pd.DataFrame(
    lignes,
    columns=data.columns
).reset_index(drop=True)

# Affichage du résultat
print(data_to_csv)
print("Nombre de lignes avant :", len(data))
print("Nombre de lignes après :", len(data_to_csv))
```
